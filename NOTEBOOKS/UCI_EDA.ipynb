```python
import pandas as pd
import numpy as np

file_path = "../Datasets/UCI/data.csv"

df = pd.read_csv(file_path, sep=";")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Display first 5 rows
df.head()
```

```python
# Display all column names and their data types
print("Number of columns:", len(df.columns))
print("\nColumn names and data types:")
print(df.dtypes)
```

```python
# Check for missing values

missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)

print("\nTotal missing values:", missing_values.sum())
```

```python
# Check the distribution of the target variable

print("Target distribution:")
print(df["Target"].value_counts())

print("\nTarget distribution in percentage:")
print(df["Target"].value_counts(normalize=True) * 100)
```

```python
import matplotlib.pyplot as plt

# Target distribution
target_counts = df["Target"].value_counts()

plt.figure(figsize=(8, 5))
target_counts.plot(kind="bar")

plt.title("Distribution of Student Academic Outcomes")
plt.xlabel("Student Outcome")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.tight_layout()

plt.show()
```

```python
# Statistical summary of numerical features

df.describe().T
```

```python
# Correlation analysis of numerical features

import matplotlib.pyplot as plt

# Select numerical columns
numeric_df = df.select_dtypes(include=["int64", "float64"])

# Calculate correlation matrix
correlation_matrix = numeric_df.corr()

# Display correlation with target-related numerical features
print(correlation_matrix.round(2))
```

```python
# Correlation heatmap

import matplotlib.pyplot as plt

plt.figure(figsize=(16, 12))

plt.imshow(correlation_matrix, cmap="coolwarm", aspect="auto")

plt.colorbar(label="Correlation")

plt.xticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns,
    rotation=90
)

plt.yticks(
    range(len(correlation_matrix.columns)),
    correlation_matrix.columns
)

plt.title("Correlation Heatmap of Numerical Features")

plt.tight_layout()
plt.show()
```

```python
# Attendance mode vs student academic outcome

import matplotlib.pyplot as plt
import pandas as pd

attendance_table = pd.crosstab(
    df["Daytime/evening attendance\t"],
    df["Target"]
)

print("Attendance mode vs student outcome:")
print(attendance_table)

attendance_table.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Attendance Mode vs Student Academic Outcome")
plt.xlabel("Attendance Mode (0 = Evening, 1 = Daytime)")
plt.ylabel("Number of Students")
plt.xticks(rotation=0)
plt.legend(title="Student Outcome")

plt.tight_layout()
plt.show()
```

```python
# Step 20: Correlation of numerical features with target

# Convert target into numerical values
target_mapping = {
    "Dropout": 0,
    "Enrolled": 1,
    "Graduate": 2
}

df["Target_encoded"] = df["Target"].map(target_mapping)

# Select numerical columns
numeric_df = df.select_dtypes(include=["int64", "float64"])

# Calculate correlation with target
target_correlation = numeric_df.corr()["Target_encoded"].sort_values(ascending=False)

print("Correlation of features with academic outcome:")
print(target_correlation)
```
